```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- File paths ---
src_path = r"C:\Users\egw\Downloads\frame_00000.jpg"   # overexposed
ref_path = r"C:\Users\egw\Downloads\frame_00788.jpg"   # good tone

src = cv2.cvtColor(cv2.imread(src_path), cv2.COLOR_BGR2RGB)
ref = cv2.cvtColor(cv2.imread(ref_path), cv2.COLOR_BGR2RGB)

# --- Click selectors ---
print("🖱️ Click on a BRIGHT area in frame_00000")
plt.imshow(src)
pts_src = plt.ginput(1)
plt.close()

print("🖱️ Click on a PREFERRED tone in frame_00788")
plt.imshow(ref)
pts_ref = plt.ginput(1)
plt.close()

if not pts_src or not pts_ref:
    raise SystemExit("❌ No points selected!")

src_y, src_x = map(int, pts_src[0])
ref_y, ref_x = map(int, pts_ref[0])
bright_color = src[src_x, src_y].astype(np.float32)
target_color = ref[ref_x, ref_y].astype(np.float32)

print(f"🎨 Bright color: {bright_color}")
print(f"🎯 Target tone:  {target_color}")

# --- Replace all bright pixels with similar colors ---
lab = cv2.cvtColor(src, cv2.COLOR_RGB2LAB)
mean_bright = cv2.cvtColor(np.uint8([[bright_color]]), cv2.COLOR_RGB2LAB)[0,0]
mean_target = cv2.cvtColor(np.uint8([[target_color]]), cv2.COLOR_RGB2LAB)[0,0]

# Create mask of bright pixels (distance in LAB space)
dist = np.linalg.norm(lab - mean_bright, axis=2)
mask = (dist < 35).astype(np.float32)  # threshold controls how wide the region is

# Apply smooth blending
mask = cv2.GaussianBlur(mask, (25, 25), 0)
shift = (mean_target - mean_bright).astype(np.float32)
corrected_lab = lab + (mask[..., None] * shift)
corrected_lab = np.clip(corrected_lab, 0, 255).astype(np.uint8)
result = cv2.cvtColor(corrected_lab, cv2.COLOR_LAB2RGB)

# --- Display ---
plt.figure(figsize=(15,5))
plt.subplot(1,3,1); plt.imshow(src); plt.title("Original Frame 00000")
plt.subplot(1,3,2); plt.imshow(mask, cmap='gray'); plt.title("Bright Area Mask")
plt.subplot(1,3,3); plt.imshow(result); plt.title("Recolored Result")
plt.show()

# --- Optional save ---
out_path = r"C:\Users\egw\Downloads\frame_00000_recolored.jpg"
cv2.imwrite(out_path, cv2.cvtColor(result, cv2.COLOR_RGB2BGR))
print(f"✅ Saved: {out_path}")
```
